## 2. Background theory

- A generator produces values on demand, so it never builds the whole sequence in memory.

### Step 2 — Generators: producing values one at a time

A generator computes each value only when it is asked for. In search problems the list of possible next moves can be enormous, so producing them one at a time matters. Note what happens when you consume a generator twice.

```python
def squares_list(n):
    return [x * x for x in range(n)]        # builds the whole list at once

def squares_gen(n):
    for x in range(n):
        yield x * x                          # produces one value at a time

print('list   :', squares_list(5))
print('gen    :', list(squares_gen(5)))

g = squares_gen(5)
print('first sum :', sum(g))
print('second sum:', sum(g), '   <- a generator is consumed once')
```

### Step 3 — An agent and its environment

Here is a minimal agent in a corridor of rooms. It can move left or right, it senses only whether the current room is dirty, and its belief state records which rooms it has already cleaned. Read the loop and identify the four stages from Lecture 2.

```python
import random
random.seed(0)

class Environment:
    def __init__(self, n_rooms=5):
        self.n = n_rooms
        self.dirty = {r: random.choice([True, False]) for r in range(n_rooms)}
        self.position = 0

    def percept(self):
        # The agent sees ONLY whether the current room is dirty.
        return {'dirty': self.dirty[self.position]}

    def act(self, action):
        if action == 'clean':
            self.dirty[self.position] = False
        elif action == 'right' and self.position < self.n - 1:
            self.position += 1
        elif action == 'left' and self.position > 0:
            self.position -= 1

env = Environment()
print('true state (hidden from the agent):', env.dirty)
print('what the agent can see           :', env.percept())
```

### Step 4 — The sense-update-decide-act loop

This agent keeps a belief state: the set of rooms it believes it has cleaned. Notice that the decision uses only the belief state and the current percept.

```python
def run_agent(env, steps=12, verbose=True):
    cleaned = set()                       # THE BELIEF STATE
    for t in range(steps):
        percept = env.percept()           # SENSE
        if percept['dirty']:              # UPDATE + DECIDE
            action = 'clean'
            cleaned.add(env.position)
        else:
            action = 'right'
        if verbose:
            print('t=%2d  room=%d  percept=%-5s  action=%-5s  belief=%s'
                  % (t, env.position, percept['dirty'], action, sorted(cleaned)))
        env.act(action)                   # ACT
    return cleaned

env = Environment()
cleaned = run_agent(env)
print('\nrooms still dirty:', [r for r, d in env.dirty.items() if d])
```

### Task 2 — Give the agent a better policy

The agent above always moves right, so it walks off the end of the corridor and stops making progress. Write a new policy that uses the belief state to turn around when it reaches the end. Run it and confirm every room is cleaned.

> **Marks:** 4

```python
def make_env(n):
    """Corridor where EVERY room is dirty (worst case), agent starts in room 0."""
    env = Environment(n)
    env.dirty = {r: True for r in range(n)}
    return env

def run_agent_v2(env, max_steps=200, verbose=True):
    # BELIEF STATE: which rooms have been visited + which way I am walking
    belief = {'direction': 'right', 'visited': set()}
    actions = 0
    for t in range(max_steps):
        percept = env.percept()                    # SENSE
        belief['visited'].add(env.position)        # UPDATE
        if percept['dirty']:                       # DECIDE
            action = 'clean'
        else:
            if len(belief['visited']) == env.n:    # every room seen and current one clean
                break
            if env.position == env.n - 1:          # right end -> turn around
                belief['direction'] = 'left'
            elif env.position == 0:                # left end  -> turn around
                belief['direction'] = 'right'
            action = belief['direction']
        if verbose:
            print('t=%2d  room=%d  dirty=%-5s  action=%-5s  visited=%s'
                  % (t, env.position, percept['dirty'], action, sorted(belief['visited'])))
        env.act(action)                            # ACT
        actions += 1
    return actions

env = make_env(5)
n_actions = run_agent_v2(env)
print('\nrooms still dirty:', [r for r, d in env.dirty.items() if d])
print('every room cleaned?', not any(env.dirty.values()), '| actions:', n_actions)
```

### Task 3 — Break the belief state deliberately

Remove the belief state from your agent — that is, stop recording which rooms have been cleaned — and run it again. Describe in a comment what changes about its behaviour, and explain why in terms of Lecture 2.

> **Marks:** 4

```python
# Agent with NO belief state: it only looks at its current position and percept.
# It cannot remember its direction or which rooms it has visited.
def run_agent_no_belief(env, steps=40, verbose=True):
    for t in range(steps):
        percept = env.percept()
        if percept['dirty']:
            action = 'clean'
        elif env.position == env.n - 1:
            action = 'left'
        else:
            action = 'right'      # in the middle it cannot know which way it was going
        if verbose and t < 14:
            print('t=%2d  room=%d  dirty=%-5s  action=%s' % (t, env.position, percept['dirty'], action))
        env.act(action)

env = make_env(5)
run_agent_no_belief(env)
print('...')
print('rooms still dirty after 40 steps:', [r for r, d in env.dirty.items() if d])
print('but the agent is STILL moving - it has no way to know it is finished')

# EXPLANATION
# What changes: the rooms do get cleaned (the agent sweeps right and cleans each dirty room
# it meets), but the behaviour afterwards is useless. It reaches the right end, turns left,
# and one step later it is in the middle again with no memory of walking left, so it goes
# right again. It bounces between the last two rooms forever, never explores the left side
# again and never knows that the job is done, so it never stops (it wastes every extra action).
# Why (Lecture 2): the agent decides from the belief state + current percept only. The belief
# state is the bounded summary of the past. If it is removed, the agent has no summary of the
# past at all, so its action depends only on the present moment. It cannot keep a direction,
# cannot tell which rooms are already done, and cannot decide "everything is clean, stop".
```

### Task 4 — Count the work

Modify your agent to count how many actions it takes to clean every room. Run it on corridors of 5, 10 and 20 rooms and record the counts in the observations table.

> **Marks:** 3

```python
def count_actions(n, max_steps=200):
    env = make_env(n)
    actions = run_agent_v2(env, max_steps=max_steps, verbose=False)
    return actions, not any(env.dirty.values())

def count_actions_no_belief(n, steps=100):
    env = make_env(n)
    run_agent_no_belief(env, steps=steps, verbose=False)
    return steps, not any(env.dirty.values())    # it never stops by itself, so we cap it

for n in (5, 10, 20):
    a, ok = count_actions(n)
    b, ok2 = count_actions_no_belief(n)
    print('%2d rooms | with belief: stops after %3d actions, all clean=%s | without belief: never stops (cut at %d), all clean=%s'
          % (n, a, ok, b, ok2))
```
